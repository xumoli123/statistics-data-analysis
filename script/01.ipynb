```python
metric = 0.032
type(metric)
# -> <class 'float'>

# 点击率
print(type(metric))
# -> <class 'float'>

metric = "现在我将浮点数变为了字符串"
print(type(metric))
# -> <class 'str'>

metric = [0.032, 0.028, 0.041]
print(type(metric))
# -> <class 'list'>
```
